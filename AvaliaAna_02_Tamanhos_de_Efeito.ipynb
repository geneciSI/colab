# Calculando os tamanhos de efeito



Este caderno tem por objetivos realizar cálculos de tamanhos de efeito com base [neste documento](https://drive.google.com/file/d/1FCxDCIJI_c0EgiiRnA8_MkSstfLJJ07p/view?usp=sharing).

As instruções a seguir carregam uma base de dados de **NOTAS** (de um curso) contendo as seguintes informações:
* Aluno (nome do aluno)
* Sexo
* NotaPre (nota do aluno no pré-teste)
* NotaPos (nota do aluno no pós-teste)

```
import pandas as pd
#https://drive.google.com/file/d/1Ao_E_lmpH3eqrMoonw7QC4HFrfAFjzXj/view?usp=sharing
codigo = "1Ao_E_lmpH3eqrMoonw7QC4HFrfAFjzXj"
file = "https://drive.google.com/u/3/uc?id=" + codigo + "&export=download"
curso = pd.read_csv(file, sep = ";")
```

```
# Exibe os dados armazenados
curso
```


As instruções a seguir calculam:
* Média de NotaPre
* Média de NotaPos
* Desvio Padrão de NotaPre
* Desvio Padrão de NotaPos
* Dispersão Média
* Tamanho de Efeito (da turma)

```
mediaPre = curso.NotaPre.mean()   # calcula a média de NotaPre
mediaPos = curso.NotaPos.mean()   # calcula a média de NotaPos
desvioPre = curso.NotaPre.std()   # calcula o desvio padrão de NotaPre
desvioPos = curso.NotaPos.std()   # calcula o desvio padrão de NotaPos
mediaDesvio = (desvioPre+desvioPos)/2     # calcula a dispersão média
TamEfeito = (mediaPos-mediaPre)/mediaDesvio  # calcula o tamanho de efeito
# as instruções a seguir mostram os resultados
print("Média NotaPre: ", mediaPre)
print("Média NotaPos: ", mediaPos)
print("Dispersão Média:", mediaDesvio)
print("Tamanho de Efeito: ", TamEfeito)
```

As instruções a seguir reexibem os resultados apenas com 2 casas decimais

```
# as instruções a seguir mostram os resultados com DUAS CASAS DECIMAIS
print("Média NotaPre: %.2f " % mediaPre)
print("Média NotaPos: %.2f" % mediaPos)
print("Dispersão Média: %.2f" % mediaDesvio)
print("Tamanho de Efeito: %.2f" % TamEfeito)
```

As instruções a seguir calculam o "tamanho de efeito" de dois alunos.

```
#Calculando o Tamanho do Efeito de um aluno (Fred)
notaPre= 50
notaPos= 75
TE_aluno = (notaPos-notaPre)/mediaDesvio  # calcula o tamanho de efeito
print("Tamanho de Efeito (Fred): %.2f" % TE_aluno)
```
