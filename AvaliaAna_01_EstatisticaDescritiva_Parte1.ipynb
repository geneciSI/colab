## Medidas de Tendência Central
A partir dos dados dos pacientes podemos querer saber como vai a saúde deste grupo. Se a lista for muito grande, torna-se complicado analisar um a um. Um primeiro passo é analisar informações que sumarizam (resumem) os dados e então verificar se é possível tirar conclusões. Em geral, podemos ter uma boa ideia dessa analisando a tendência central.

Uma **tendência central** (ou, normalmente, uma medida de tendência central) é um valor central ou valor típico para um certo atributo. As medidas  de tendência central mais comuns são:

- **Média aritmética** (ou simplesmente, média) - obtida pela soma de todas as medições divididas pelo número de observações no conjunto de dados.
- **Mediana** - obtida a partir dos dados ordenados, encontrando-se o valor (pertencente ou não à amostra) que a divide ao meio, isto é, 50% dos elementos da amostra são menores ou iguais à mediana e os outros 50% são maiores ou iguais à mediana.
- **Moda** -  valor que aparece com maior frequência no conjunto de dados.

O Pandas tem funções pré-definidas que calculam essas medidas de tendência central:

### Média

Por exemplo, para calcular a média das avaliações, podemos usar a função `mean`, que calcula a média de cada coluna da base de dados.

````
##  média aritimética
curso.mean()
````

A instrução acima, aplicada ao `dataframe` (tabela de dados) mostra a média de todas os atributos. Mas por exemplo, não faz sentido calcular a média do atributo "Sexo".

Além disso, normalmente queremos analisar a média de um determinado atributo que queremos investigar.



#### Média especificando o atributo

Para calcular a média de um atributo específico, como a "NotaPre" (préteste), devemos especificar o nome do atributo.

Pode-se usar essa notação a seguir, com o atributo entre colchetes: **`[NotaPre]`**

````
#média de NotaPre
curso["NotaPre"].mean()
````

Outra notação, um pouco mais simples é usar o nome do atributo direto (se os colchetes, mas com um ponto extra): **`.NotaPre`**

````
# Outra possibilidade de obtenção da média de um determinado atributo
curso.NotaPre.mean()
````

O Python nos permite restringir o número de casas decimais, usando o print com formatação. Veja o exemplo a seguir. Note que existe um espaço em

````
# imprimindo com 2 casas decimais: .2
print("%.2f" % curso.NotaPre.mean())
````

Outra possibilidade é atribuir o resultado para uma variável (por exemplo, media) antes de mostrar na tela (com o `print`)

````
# atribuindo a média para uma variável
media = curso.NotaPre.mean()
# imprimindo com 2 casas decimais: .2
print("Média da NotaPre: %.2f" % media)
````

Para tentar analisar o significado dessa média, podemos verificar os valores extremos de `NotaPre`(maior e menor).

````
# Maior Valor NotaPre
curso.NotaPre.max()
````

````
# Menor Valor de NotaPre
curso.NotaPre.min()
````

Nota-se uma variação razoável entre a média `mean()` (48.5) , a maior `max()` NotaPre (70) e a menor `min()` NotaPre (20). Poderemos entender melhor esta variação, mais adiante, com as medidas de dispersão. Mas antes disso, vamos ver as outras medidas de tendência central.

### Mediana

A mediana pode ser calculada usando a função `median`:

````
curso.median()
````

#### Mediana especificando o atributo
Calculando a mediana somente do atributo `Idade`

````
curso.NotaPre.median()
````

Nesse caso, os valores da média e da mediana (de NotaPre) diferem um pouco (respectivamente 48,50 e 52.5), e ambos poderiam ser usados como um representante da tendência central do atributo `NotaPre`.  Vale lembrar, que mesmo havendo valores extremos de `NotaPre` (mínimo = 20 e máximo = 70) razoavelmente distantes da Média (48.5), ainda assim a Média e Mediana são valores "relativamente" próximos.

Na presença de valores extremos, a mediana pode ser uma medida **menos sensível** a presença desses valores.



### Moda

A moda representa o valor que mais ocorre na base de dados. A moda pode não existir o não ser única.

Mas o Pandas, possui a função `mode` para calcular a moda:

````
# moda
curso.NotaPre.mode()
````

A instrução acima, mostrou a moda do atributo `NotaPre`

Caso haja mais de uma moda, todos os valores serão exibidos. Veja o caso de `NotaPos` a seguir que tem "duas modas": 30 e 75.

````
# moda
curso.NotaPos.mode()
````

### Exibindo Média, Mediana e Moda com print

Veja a seguir que podemos usar o print para exibir uma mensagem com um RÓTULO, indicando o que significa o valor exibido.

````
print("Média de NotaPre :", curso.NotaPre.mean())
print("Mediana de NotaPre: ", curso.NotaPre.median())
print("Moda de NotaPre: ", curso.NotaPre.mode())
````

## Medidas de dispersão

A **medidas de dispersão** nos ajudam a entender como os dados variam em torno da média, ou seja, também ajudam a descrever a tendência nos dados.

Três medidas de dispersão normalmente usadas são:

- a **amplitude**, que consiste na diferença entre o maior e o menor valor da série.
- a **variância**, que consiste na média do quadrado diferença entre cada valor do atributo e a sua média. Isto parece meio "confuso" mas vamos ver uma explicação mais detalhada adiante.
- o **desvio padrão**, que é a raiz quadrada da variância. Isto também não é muito esclarecedor mas vamos ver uma explicação mais detalhada adiante.


### Amplitude

No pandas **não** temos um método para calcular a amplitude, mas podemos facilmente calcular fazendo a diferença entre a nota máxima e a mínima:

````
# amplitude
print("NotaPre máxima: ",curso.NotaPre.max())
print("NotaPre mínima: ",curso.NotaPre.min())
amplitude = curso.NotaPre.max() - curso.NotaPre.min()
print("Amplitude NotaPre: ", amplitude)
````

Para ter uma idéia do que significa essa amplitude, podemos exibir o valor da média do atributo NotaPre.

````
print("NotaPre Média: ",curso.NotaPre.mean())
````

Tomando como exemplo o atributo `NotaPre`, a amplitude nesse conjunto de dados é igual a 50.0, ou seja, a diferença entre a maior e a menor `NotaPre`.  

Note que este valor, 50.0, é muito alto, uma vez que a média é 48.50.

Precisamos analisar outras medidas de dispersão como variância e desvio padrão.

### Variância

Para calcular a variância, podemos usar a função `var` do pandas:

````
curso.var()
````

## Desvio Padrão

Para para calcular o desvio padrão podemos usar a função `std()` do pandas.
É importante destacar que  o pandas (por meio de `std()`) calcula o desvio padrão da **AMOSTRA**, geralmente mais utilizado. Para obtenção do desvio padrão da população há opções na biblioteca `numpy`.

````
#desvio padrão
curso.std()
````

A variância e o desvio padrão mostram, de certa forma, como os valores estão dispersos em torno na média.

O desvio padrão nos dá uma ideia de qual é a diferença média (variação) entre todos os pontos (cada valor) e a média desses pontos.

Já a variância representa o quadrado dessa diferença média.

Para tentarmos entender o que significa e para que serve a variância e o desvio padrão, vamos analisar um determinado atributo, como por exemplo a NotaPre.

````
print("Quantidade: ", curso.NotaPre.count())
print("Medidas de Tendência Central =============")
print("Média:", curso.NotaPre.mean())
print("Mediana:", curso.NotaPre.median())
print("Moda:", curso.NotaPre.mode())
print("Medidas de Dispersão ==================")
maximo= curso.NotaPre.max()
minimo= curso.NotaPre.min()
print("Amplitude :", (maximo - minimo), " Máximo: ", maximo, " Mínimo: ", minimo)
print("Variância : ", curso.NotaPre.var())
print("Desvio Padrão: ", curso.NotaPre.std())
````

Analisando as medidas de tendência central e de dispersão de pressão Sistólica, temos que:

média = 48.5
menor = 20 (28.5 abaixo da média)
maior = 70 (21.5 acima da média)
Amplitude = 50
Já o desvio padrão (desvio médio) é 16.33, indicando que em média, cada valor de `NotaPre` varia (para mais ou para menos, 16.33 em relação a média aritmética (48.5).

Este desvio parece indicar que os dados (de NotaPre) são um tanto "distantes" um do outro (variam "muito").

### Várias Medidas

Assim, para uma análise mais apurada deve-se olhar várias medidas.

O pandas tem uma função que calcula diferentes medidas de tendência central e dispersão em uma única instrução, que é a describe():

````
curso.NotaPre.describe()
````

Essa função mostra a contagem (`count`), que é o número de valores não nulos, a média (`mean`), desvio padrão (`std`) e mediana (`50%`). Além disso, também são mostrados o mínimo (`min`) e máximo (`max`), e o 1o. (`25\%` - $Q_1$) e 3o. (`75%` - $Q_3$) quartis.  Assim como a mediana divide os valores ao meio (por isso `50%`) e utilizamos o valor que está bem nesse meio, para os quartis dividimos a série em quatro partes, e utilizamos o valor entre o 1a. e 3a. parte para representar os quartis.
